# Optional lab 3 — Build a gas network in pandapipes

**Optional extension · 15 minutes · requires `pip install pandapipes`**

This is an independent model using the **pandapipes** physical-property library and numerical solver, **not** a like-for-like validation of the customized Darcy solver in Lab 1. The fluids, parameterizations, and friction assumptions may differ.

Documentation: https://www.pandapipes.org/start/ and https://pandapipes.readthedocs.io/en/latest/.

```python
try:
    import pandapipes as pp
except ImportError:
    pp=None
    print('pandapipes not installed. In a terminal run: pip install pandapipes')
```

```python
if pp is not None:
    net=pp.create_empty_network(fluid='lgas')
    j0=pp.create_junction(net,pn_bar=5.0,tfluid_k=293.15,name='source')
    j1=pp.create_junction(net,pn_bar=5.0,tfluid_k=293.15,name='junction')
    j2=pp.create_junction(net,pn_bar=5.0,tfluid_k=293.15,name='consumer A')
    j3=pp.create_junction(net,pn_bar=5.0,tfluid_k=293.15,name='consumer B')
    pp.create_ext_grid(net,junction=j0,p_bar=5.5,t_k=293.15)
    pp.create_sink(net,junction=j2,mdot_kg_per_s=0.03)
    pp.create_sink(net,junction=j3,mdot_kg_per_s=0.02)
    pp.create_pipe_from_parameters(net,from_junction=j0,to_junction=j1,length_km=0.5,diameter_m=0.11)
    pp.create_pipe_from_parameters(net,from_junction=j1,to_junction=j2,length_km=0.7,diameter_m=0.09)
    pp.create_pipe_from_parameters(net,from_junction=j1,to_junction=j3,length_km=0.9,diameter_m=0.09)
    pp.pipeflow(net)
    print(net.res_junction[['p_bar']])
    print(net.res_pipe[['mdot_from_kg_per_s']])
```
